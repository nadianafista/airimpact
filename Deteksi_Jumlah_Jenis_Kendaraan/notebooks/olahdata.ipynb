```python
# load csv dalam data frame
df = pd.read_csv(r'C:\Users\NADIA\OneDrive\ドキュメント\airimpact\Deteksi_Kendaraan\Deteksi_Jumlah_Jenis_Kendaraan\data\data_bersih.csv')
```

```python
# ganti semua data ke string

df['detection_timestamp'] = df['detection_timestamp'].astype(str)
```

```python
# ganti value dari 58200 jadi 2026
df['detection_timestamp'] = df['detection_timestamp'].str.replace('58200', '2026')
```

```python
df.head(10)
```

```python
# Konversi 'timestamp' ke datetime terlebih dahulu (ini tanggal yang valid)
df['timestamp'] = pd.to_datetime(df['timestamp'], format='ISO8601')
```

```python
# Ambil TANGGAL dari 'timestamp' (Jadikan string dengan format YYYY-MM-DD)
tanggal_benar = df['timestamp'].dt.strftime('%Y-%m-%d')
```

```python
# FIX: format aslinya '2026-03-25 13:46:40' (pakai SPASI, bukan 'T'),
# jadi split('T') menghasilkan NaN -> semua jadi NaT.
# Pakai regex supaya jam:menit:detik terambil, apa pun pemisahnya (spasi atau 'T').
waktu_benar = df['detection_timestamp'].astype(str).str.extract(r'(\d{1,2}:\d{2}:\d{2}(?:\.\d+)?)')[0]
```

```python
df['detection_timestamp'] = pd.to_datetime(tanggal_benar + ' ' + waktu_benar, errors='coerce')

# Cek: harusnya 0. Kalau masih ada, baris yang gagal diparse ditampilkan di bawah.
jumlah_nat = df['detection_timestamp'].isna().sum()
print(f"Jumlah detection_timestamp yang gagal diparse (NaT): {jumlah_nat}")
if jumlah_nat > 0:
    print(df.loc[df['detection_timestamp'].isna()].head())
    df = df.dropna(subset=['detection_timestamp'])
```

```python
print(df['detection_timestamp'].dtype)
print(df['timestamp'].dtype)

print(df[['detection_timestamp', 'timestamp']].head())
```

```python
df['timestamp'] = (
    df['timestamp']
    .dt.tz_localize('UTC')
    .dt.tz_convert('Asia/Jakarta')
)
```

```python
# Convert detection_timestamp to Asia/Jakarta
df['detection_timestamp'] = (
    df['detection_timestamp']
    .dt.tz_localize('UTC')
    .dt.tz_convert('Asia/Jakarta')
)
```

```python
df['detection_timestamp'] = df['detection_timestamp'].dt.tz_localize(None)
df['timestamp'] = df['timestamp'].dt.tz_localize(None)
```

```python
print(df[['detection_timestamp', 'timestamp']].head())
```

```python
df.head(10)
```

```python
# Kolom lokasi tidak dipakai, jadi dihapus
df = df.drop(columns=['location'], errors='ignore')
```

```python
df.head(10)
```

```python
# Kita hitung jumlah track_id yang unik (nunique)
df_harian = df.resample('D', on='detection_timestamp')['track_id'].nunique().reset_index()
```

```python
# 2. Rapikan nama kolomnya agar enak dibaca
df_harian.columns = ['Tanggal', 'Total_Kendaraan']
```

```python
# 3. Urutkan data dari yang paling banyak ke paling sedikit (Descending)
df_harian_tertinggi = df_harian.sort_values(by='Total_Kendaraan', ascending=False)
```

```python
# 4. Tampilkan hasilnya (misal 5 hari paling ramai)
print("--- 5 Hari dengan Volume Kendaraan Tertinggi ---")
print(df_harian_tertinggi.head())
```

```python
# Ambil otomatis tanggal paling ramai dari hasil df_harian_tertinggi
tanggal_puncak = df_harian_tertinggi.iloc[0]['Tanggal'].strftime('%Y-%m-%d')
df_puncak = df[df['detection_timestamp'].dt.strftime('%Y-%m-%d') == tanggal_puncak]
```

```python
print(f"Total baris data di tanggal {tanggal_puncak}: {len(df_puncak)}")
df_puncak.head()
```

```python
# Kelompokkan data df_puncak per jam ('H')
df_puncak_per_jam = df_puncak.resample('h', on='detection_timestamp')['track_id'].nunique().reset_index()
df_puncak_per_jam.columns = ['Waktu', 'Volume_Kendaraan']
```

```python
# Tampilkan 5 jam paling sibuk di hari itu
jam_sibuk = df_puncak_per_jam.sort_values(by='Volume_Kendaraan', ascending=False)
print(f"--- Jam Paling Sibuk di Tanggal {tanggal_puncak} ---")
print(jam_sibuk.head())
```

```python
# Menghitung jenis kendaraan khusus di hari itu
komposisi_kendaraan = df_puncak.groupby('object')['track_id'].nunique().sort_values(ascending=False).reset_index()
print("\n--- Komposisi Kendaraan ---")
print(komposisi_kendaraan)
```

## Rekap Akhir: Jumlah Kendaraan per Jenis (seluruh data)


```python
# Satu kendaraan bisa muncul di banyak baris (banyak frame), jadi yang dihitung
# adalah kombinasi unik camera_id + track_id, bukan jumlah baris.
rekap_jenis = (
    df.drop_duplicates(subset=['camera_id', 'track_id', 'object'])
      .groupby('object')
      .size()
      .sort_values(ascending=False)
      .reset_index(name='Jumlah_Kendaraan')
      .rename(columns={'object': 'Jenis_Kendaraan'})
)
rekap_jenis['Persentase'] = (rekap_jenis['Jumlah_Kendaraan'] / rekap_jenis['Jumlah_Kendaraan'].sum() * 100).round(2)

print("=== JUMLAH KENDARAAN TERDETEKSI PER JENIS ===")
print(rekap_jenis.to_string(index=False))
print(f"\nTOTAL KENDARAAN: {rekap_jenis['Jumlah_Kendaraan'].sum()}")
```

```python
# (Opsional) Rincian per jenis kendaraan per hari
rekap_jenis_harian = (
    df.assign(Tanggal=df['detection_timestamp'].dt.date)
      .drop_duplicates(subset=['Tanggal', 'camera_id', 'track_id', 'object'])
      .pivot_table(index='Tanggal', columns='object', values='track_id', aggfunc='count', fill_value=0)
)
rekap_jenis_harian['TOTAL'] = rekap_jenis_harian.sum(axis=1)
print(rekap_jenis_harian)
```
